# Name Entity Recognition (NER)

**Named Entity Recognition (NER)** is a subtask of information extraction that seeks to locate and classify named entities in text into predefined categories such as person names, organizations, locations, dates, and more.

## Key Concepts

- **Entity**: A real-world object or concept that is referred to in text (e.g., "Barack Obama", "Microsoft", "Paris").
- **Named Entity**: A specific type of entity that has a proper name.
- **NER Task**: Identifying and categorizing these entities in unstructured text.

## Common Entity Types

- **Person (PER)**: Names of people (e.g., "Elon Musk")
- **Organization (ORG)**: Companies, institutions, etc. (e.g., "Google")
- **Location (LOC)**: Geographical locations (e.g., "London")
- **Miscellaneous (MISC)**: Other entities (e.g., "COVID-19", "iPhone")


```python
str = "Barack Obama visited Microsoft headquarters in Paris on July 14, 2020."
```

```python
import nltk
nltk.download('maxent_ne_chunker_tab')
nltk.download('words')
```

```python
words = nltk.word_tokenize(str)
tagged_words = nltk.pos_tag(words)
nltk.ne_chunk(tagged_words)
```

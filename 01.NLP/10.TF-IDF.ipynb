# 🧠 TF-IDF (Term Frequency – Inverse Document Frequency)

## 📌 What is TF-IDF?

**TF-IDF** is a numerical statistic used to measure how *important* a word is in a document **relative to a collection of documents (corpus)**.

It improves on Bag of Words (BoW) by not only counting words but also **down-weighting very common words** and **up-weighting rare but meaningful words**.

---

## 🧩 Why do we need TF-IDF?

BoW has a big limitation:

- Words like **"the", "is", "and", "this"** appear in almost every text.
- So they get **high counts** even though they carry **little meaning**.

TF-IDF fixes this by:

- Keeping high scores for **important** words (rare in corpus but frequent in one document)
- Giving low scores to **common** words across many documents

---

# 🔍 TF-IDF = TF × IDF

TF-IDF is the product of two values:

---

## 1️⃣ **Term Frequency (TF)**  
TF shows **how frequently a word appears in a document**.

### Formula  
$$TF(t, d) = \frac{\text{Number of times term } t \text{ appears in document } d}{\text{Total words in document } d}$$

### Intuition
- Higher TF → term appears more often → important for that document  
- Normalized TF avoids giving longer documents unfair advantage

---

## 2️⃣ **Inverse Document Frequency (IDF)**  
IDF measures **how rare** a word is across all documents.

### Formula  
$$IDF(t) = \log\left(\frac{N}{df_t}\right)$$

Where:  
- **N** = total number of documents  
- **df_t** = number of documents containing term *t*

### Intuition  
- If a word appears in **every** document → IDF ≈ 0 → not important  
- If a word appears in **few** documents → IDF is high → meaningful

---

# 🔥 TF-IDF Intuition Summary

| Word Example | TF | IDF | TF-IDF Meaning |
|--------------|----|-----|----------------|
| **"computer"** appears frequently in only *one* document | high | high | very important |
| **"the"** appears in all documents | high | **low** | not important |
| **Rare word** appearing once in one doc | low TF | high IDF | medium importance |

---

# 📊 Example to Understand TF-IDF

### Corpus
1. "NLP is fun"
2. "NLP is very fun"
3. "NLP and ML are related"

### Word: **"fun"**
- Appears 3 times (2 docs)
- TF doc1 = 1/3  
- IDF = log(3/2)  
- TF-IDF = (1/3) × log(3/2)

### Word: **"nlp"**
- Appears in all 3 docs → common
- IDF = log(3/3) = log(1) = **0**
- TF-IDF = always 0  
➡️ TF-IDF automatically lowers weight of common words

---

# 🎯 Why TF-IDF is Useful

### ✔️ Highlights important words  
Like **"urgent", "payment", "bitcoin", "error"**

### ✔️ Reduces noise  
Common words ("is", "the", "and") automatically get low weight

### ✔️ Works very well for:
- Information retrieval  
- Search engines  
- Keyword extraction  
- Text classification (spam, sentiment, topics)

### ✔️ Still interpretable  
Unlike neural embeddings, TF-IDF scores are easy to understand

---

# ❌ Limitations of TF-IDF

| Issue | Explanation |
|--------|-------------|
| ❌ Ignores word order | "not good" and "good not" treated same |
| ❌ Ignores semantics | "happy" and "joyful" unrelated |
| ❌ Sparse vectors | High-dimensional like BoW |
| ❌ Sensitive to preprocessing | Stopwords strongly influence results |

---

# 🧠 TF-IDF vs Bag of Words (BoW)

| Feature | Bag of Words | TF-IDF |
|---------|--------------|--------|
| Counts word frequency | ✔️ Yes | ✔️ Yes |
| Downweights common words | ❌ No | ✔️ Yes |
| Highlights unique terms | ❌ No | ✔️ Yes |
| Useful for search relevance | ❌ No | ✔️ Yes |
| Easy to understand | ✔️ Yes | ✔️ Yes |

---

```python
import pandas as pd
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from sklearn.feature_extraction.text import TfidfVectorizer
# NLTK downloads
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
df = pd.DataFrame({
    "text": [
        "Congrats! You have won a $500 Walmart gift card. Click here to claim.",
        "Hey John, are we still meeting tomorrow?",
        "Lowest price guaranteed! Buy now",
        "Your OTP is 234987. Do not share with anyone.",
        "Free entry in 2 a weekly competition to win tickets!",
        "Let's catch up over coffee this evening.",
        "URGENT! Your account has been compromised. Login immediately.",
        "Hi Mom, I’ll call you later."
    ],
    "label": ["spam","ham","spam","ham","spam","ham","spam","ham"]
})

# Preprocessing setup
stop_words = set(stopwords.words("english"))
lemmatizer = WordNetLemmatizer()

# Preprocessing (line-by-line)
df["clean_text"] = df["text"].str.lower()
df["clean_text"] = df["clean_text"].apply(lambda x: re.sub(r'[^a-z\s]', '', x))
df["clean_text"] = df["clean_text"].apply(lambda x: " ".join([
    lemmatizer.lemmatize(word) for word in x.split() if word not in stop_words
]))
```

```python
tfidf = TfidfVectorizer(ngram_range=(1,3), max_features=10)
X = tfidf.fit_transform(df["clean_text"])
```

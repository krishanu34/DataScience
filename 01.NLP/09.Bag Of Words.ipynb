# 🧠 Bag of Words (BoW) in NLP

| **Section** | **Details** |
|--------------|-------------|
| **Definition** | **Bag of Words (BoW)** is a text representation technique that converts documents into fixed-length numeric vectors based on **word frequency**. It ignores grammar and order, but counts how often each word appears. |
| **Concept** | Each document is treated as a “bag” (multiset) of words — order doesn’t matter, only **which words appear** and **how many times** they appear. |
| **Example Corpus** | `["NLP is fun", "Machine learning is fun, NLP is fun", "NLP and ML are related"]` |
| **Vocabulary Extracted** | `["and", "fun", "is", "learning", "machine", "ml", "nlp", "related"]` |
| **Document-Term Matrix (Word Counts)** | <table><tr><th>Word</th><th>Doc 1<br>("NLP is fun")</th><th>Doc 2<br>("Machine learning is fun, NLP is fun")</th><th>Doc 3<br>("NLP and ML are related")</th></tr><tr><td>and</td><td>0</td><td>0</td><td>1</td></tr><tr><td>fun</td><td>1</td><td><b>2</b></td><td>0</td></tr><tr><td>is</td><td>1</td><td><b>2</b></td><td>0</td></tr><tr><td>learning</td><td>0</td><td>1</td><td>0</td></tr><tr><td>machine</td><td>0</td><td>1</td><td>0</td></tr><tr><td>ml</td><td>0</td><td>0</td><td>1</td></tr><tr><td>nlp</td><td>1</td><td><b>1</b></td><td>1</td></tr><tr><td>related</td><td>0</td><td>0</td><td>1</td></tr></table> |
| **Output Example (Count-based BoW)** | ```<br>Vocabulary: ['and', 'fun', 'is', 'learning', 'machine', 'ml', 'nlp', 'related']<br><br>Document-Term Matrix (Frequencies):<br>[[0 1 1 0 0 0 1 0]<br> [0 2 2 1 1 0 1 0]<br> [1 0 0 0 0 1 1 1]]<br>``` |
| **Advantages ✅** | 1. Simple and easy to compute.<br>2. Captures **word frequency** information (unlike one-hot encoding).<br>3. Works well for classical ML models (Naive Bayes, Logistic Regression, SVM). |
| **Disadvantages ❌** | 1. **Ignores word order and context**.<br>2. **High dimensionality** for large vocabularies.<br>3. **Sparse matrix** — most values are zero.<br>4. Cannot capture **semantic similarity** (e.g., “happy” ≠ “joyful”). |
| **When to Use** | ✅ For **text classification**, **topic modeling**, or **spam detection** where word frequency matters.<br>❌ Avoid for deep NLP tasks that require meaning or context understanding. |

---

## 🔢 Binary BoW vs Normal (Count-based) BoW

| **Type** | **Description** | **Representation Example (Using Same Corpus)** | **When to Use** | **Advantages** | **Disadvantages** |
|-----------|----------------|-----------------------------------------------|------------------|----------------|--------------------|
| **Normal BoW (Count-based)** | Each entry stores the **frequency** of a word in the document. | <br>**Output:**<br>[[0 1 1 0 0 0 1 0]<br>[0 2 2 1 1 0 1 0]<br>[1 0 0 0 0 1 1 1]] | When **word frequency** carries meaning (e.g., repetition indicates emphasis). | - Captures **how often** words appear.<br>- Better for **topic modeling** or **sentiment analysis**. | - Sensitive to document length.<br>- Common words may dominate. |
| **Binary BoW** | Each entry is **1 if the word appears**, otherwise **0**, ignoring frequency. | <br>**Output:**<br>[[0 1 1 0 0 0 1 0]<br>[0 1 1 1 1 0 1 0]<br>[1 0 0 0 0 1 1 1]] | When **presence/absence** of words matters (e.g., keyword spotting, spam detection). | - Simpler and less noisy.<br>- Reduces impact of overly frequent words. | - **Loses frequency info** — cannot tell repetition or emphasis. |

# 🔡 N-Grams in NLP

| **Section** | **Details** |
|--------------|-------------|
| **Definition** | An **n-gram** is a contiguous sequence of **n items (usually words or characters)** from a given text. N-grams help capture **local word order and context** that Bag of Words (BoW) misses. |
| **Concept** | Instead of looking at single words (unigrams), we look at groups of words: <br> - **Unigram (1-gram)** → one word at a time <br> - **Bigram (2-gram)** → two consecutive words <br> - **Trigram (3-gram)** → three consecutive words, and so on. |
| **Formula** | For a given text of length *m*, the total number of n-grams = *m − n + 1* |
| **Example Sentence** | `"NLP is fun and powerful"` |
| **Generated N-grams** | <table><tr><th>Type</th><th>n</th><th>Examples</th></tr><tr><td>Unigram</td><td>1</td><td>["NLP", "is", "fun", "and", "powerful"]</td></tr><tr><td>Bigram</td><td>2</td><td>["NLP is", "is fun", "fun and", "and powerful"]</td></tr><tr><td>Trigram</td><td>3</td><td>["NLP is fun", "is fun and", "fun and powerful"]</td></tr></table> |
| **N-Gram Representation (Example Corpus)** | Suppose we have: <br>`["NLP is fun", "NLP is very fun"]` <br><br>**Unigrams:** ["nlp", "is", "fun", "very"] <br>**Bigrams:** ["nlp is", "is fun", "is very", "very fun"] |
| **Output Example (Bigram Frequencies)** | ```<br>Vocabulary: ['is fun', 'is very', 'nlp is', 'very fun']<br><br>Matrix:<br>[[1 0 1 0]<br>[0 1 1 1]]<br>``` |
| **Why Use N-grams?** | - Capture **context and word order**.<br>- Identify **common phrases** (“machine learning”, “New York”).<br>- Improve text understanding beyond single words. |
| **Advantages ✅** | 1. Preserves **local word sequence** (unlike BoW).<br>2. Helps in **phrase detection** and context understanding.<br>3. Useful in **language modeling**, **speech recognition**, and **autocomplete**. |
| **Disadvantages ❌** | 1. **Higher dimensionality** as *n* increases (combinatorial explosion).<br>2. **Sparsity** — many possible n-grams never appear.<br>3. **Limited context** — still doesn’t capture long-range dependencies. |
| **When to Use** | ✅ When **word order** and short context are important (e.g., “New York” ≠ “York New”).<br>❌ Avoid very large *n* (≥4) unless you have **large datasets**. |
| **Common Practice** | Use **unigrams + bigrams** or **bigrams + trigrams** for most NLP models.<br>Many vectorizers (like `CountVectorizer` or `TfidfVectorizer`) support this via `ngram_range=(1,2)` or `(1,3)`. |

---

### 🧠 Summary

| **N-gram Type** | **Captures** | **Example for “NLP is fun”** |
|------------------|---------------|-------------------------------|
| **Unigram (1-gram)** | Individual words | ["NLP", "is", "fun"] |
| **Bigram (2-gram)** | Adjacent word pairs | ["NLP is", "is fun"] |
| **Trigram (3-gram)** | Three-word phrases | ["NLP is fun"] |

---

### 📊 Intuition
- **BoW** treats “NLP is fun” and “fun is NLP” as identical.  
- **N-grams** understand the difference — order matters!

---

### 🚀 Real-world Applications
- **Search engines:** autocomplete and query prediction  
- **Speech recognition:** next-word prediction  
- **Text classification:** detecting phrases like “not good” vs “good”

````python
import pandas as pd
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from sklearn.feature_extraction.text import CountVectorizer
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
````

````python
df = pd.DataFrame({
    "text": [
        "Congrats! You have won a $500 Walmart gift card. Click here to claim.",
        "Hey John, are we still meeting tomorrow?",
        "Lowest price guaranteed! Buy now",
        "Your OTP is 234987. Do not share with anyone.",
        "Free entry in 2 a weekly competition to win tickets!",
        "Let's catch up over coffee this evening.",
        "URGENT! Your account has been compromised. Login immediately.",
        "Hi Mom, I’ll call you later."
    ],
    "label": ["spam","ham","spam","ham","spam","ham","spam","ham"]
})
````

````python
stop_words = set(stopwords.words("english"))
lemmatizer = WordNetLemmatizer()
````

````python
df["clean_text"] = df["text"].str.lower()
df["clean_text"] = df["clean_text"].apply(lambda x: re.sub(r'[^a-z\s]', '', x))
df["clean_text"] = df["clean_text"].apply(lambda x: " ".join([
    lemmatizer.lemmatize(word) for word in x.split() if word not in stop_words
]))
````

````python
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(df["clean_text"])
X.toarray()
````

````python
vectorizer.vocabulary_
````

````python
vectorizer2 = CountVectorizer(max_features=5)
X = vectorizer2.fit_transform(df["clean_text"])
X.toarray()
````

````python
vectorizer2.vocabulary_
````

````python
vectorizer3 = CountVectorizer(ngram_range=(1,3))
X = vectorizer3.fit_transform(df["clean_text"])
vectorizer3.vocabulary_
````

#### 🧠 One-Hot Encoding — Summary Notes

| **Section** | **Details** |
|--------------|-------------|
| **Definition** | One-hot encoding represents categorical values (like words) as binary vectors — one element is `1` (active), and all others are `0`. |
| **Example Vocabulary** | `["hello", "world", "nlp", "is", "fun"]` |
| **Example Encoding Table** | <table><tr><th>Word</th><th>One-hot Vector</th></tr><tr><td>hello</td><td>[1, 0, 0, 0, 0]</td></tr><tr><td>world</td><td>[0, 1, 0, 0, 0]</td></tr><tr><td>nlp</td><td>[0, 0, 1, 0, 0]</td></tr><tr><td>is</td><td>[0, 0, 0, 1, 0]</td></tr><tr><td>fun</td><td>[0, 0, 0, 0, 1]</td></tr></table> |
| **Python Example (Using scikit-learn)** | ```python<br>from sklearn.preprocessing import OneHotEncoder<br>import numpy as np<br><br>words = np.array(["hello", "world", "nlp", "is", "fun"]).reshape(-1, 1)<br>encoder = OneHotEncoder(sparse_output=False)<br>one_hot = encoder.fit_transform(words)<br><br>for word, vector in zip(words.flatten(), one_hot):<br>    print(f"{word:>5} -> {vector}")<br>``` |
| **Manual Implementation (No Library)** | ```python<br>words = ["hello", "world", "nlp", "is", "fun"]<br>vocab = {word: i for i, word in enumerate(words)}<br><br>def one_hot_encode(word):<br>    vector = [0] * len(vocab)<br>    vector[vocab[word]] = 1<br>    return vector<br><br>print(one_hot_encode("nlp"))<br># Output: [0, 0, 1, 0, 0]<br>``` |
| **Advantages ✅** | 1. Simple and easy to understand.<br>2. No assumptions about relationships between categories.<br>3. Deterministic — consistent representation every time. |
| **Disadvantages ❌** | 1. Produces **sparse, high-dimensional vectors** for large vocabularies.<br>2. **No semantic meaning** — all words are orthogonal.<br>3. **Fixed vocabulary** — cannot handle unseen words.<br>4. Not efficient for deep NLP models. |
| **When to Use** | ✅ Small vocabularies or categorical data.<br>❌ Large text corpora or semantic NLP tasks. |
| **Next Steps** | Learn about **Word Embeddings** (Word2Vec, GloVe, TF-IDF, BERT), which produce **dense vectors** that capture **semantic relationships** between words. |


````python
import re
import numpy as np
from sklearn.preprocessing import OneHotEncoder
````

````python
sentences = [
    "NLP is fun!",
    "Machine learning is powerful",
    "NLP and ML are related"
]
````

````python

# Step 2: Basic cleaning
cleaned = [re.sub(r'[^a-zA-Z\s]', '', s.lower()) for s in sentences]
cleaned
````

````python
# Step 3: Tokenization
tokens = [s.split() for s in cleaned]
tokens
````

````python
# Step 4: Build vocabulary
vocab = sorted(list(set([word for sent in tokens for word in sent])))
vocab
````

````python
# Step 5: One-hot encode vocabulary
encoder = OneHotEncoder(sparse_output=False)
encoded_vocab = encoder.fit_transform(np.array(vocab).reshape(-1, 1))
````

````python
# Step 6: Encode words from a sentence
for word, vec in zip(vocab, encoded_vocab):
    print(f"{word:>10} -> {vec}")
````
